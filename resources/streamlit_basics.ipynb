```python
import streamlit as st

st.title("Streamlit Basics Example")
st.header("This is a header")
st.subheader("This is a subheader")
st.text("This is a simple text element.")
st.write("This is a write element that can display various data types.")
st.markdown("**This is a markdown element with bold text.**")
user_name = st.text_input("Enter your name:")
user_phone = st.number_input("Enter your phone number:")
device_type = st.selectbox("Select your device type:", ["Mobile", "Tablet", "Desktop"])
st.checkbox("I agree to the terms and conditions.")
st.radio("Choose your preferred contact method:", ["Email", "Phone", "SMS"])
st.multiselect("Select your favorite programming languages:", ["Python", "JavaScript", "Java", "C++", "Ruby"])
st.slider("Select your age:", 0, 100, 25)
st.date_input("Select your date of birth:")
st.time_input("Select your preferred contact time:")
st.file_uploader("Upload your profile picture:")
st.color_picker("Pick your favorite color:")
st.progress(50)
st.success("This is a success message!")
st.error("This is an error message!")
st.warning("This is a warning message!")
st.info("This is an informational message!")
st.button("Submit")
st.balloons()
st.sidebar.title("Sidebar Example")
st.sidebar.text("This is a sidebar text element.")
```
