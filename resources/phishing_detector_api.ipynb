# 🔍 Fraud Message Detector
This notebook connects to the fraud detection API and checks whether a message is fraudulent.

**API Endpoint:** `https://muhdauwal-ph.hf.space/predict`

## 📦 Step 1 — Install & Import Dependencies

```python
import requests
```

## ⚙️ Step 2 — Configuration

```python
# API URL — change this if you deploy to a different server
API_URL = "https://muhdauwal-ph.hf.space/predict"
```

## 🔧 Step 3 — Define the Prediction Function

```python
def predict(text: str) -> dict:
    """
    Sends a message to the fraud detection API and returns the result.
    
    Args:
        text (str): The message to check.
    
    Returns:
        dict: API response containing status, fraud_type, and fraud_desc.
    """
    if not text.strip():
        print("⚠️  Please enter some text.")
        return {}

    print("Predicting... ⏳")

    try:
        response = requests.post(
            API_URL,
            json={"text": text}   # requests handles JSON + headers automatically
        )
        response.raise_for_status()  # raises an error for bad HTTP status (4xx, 5xx)
        data = response.json()

        # ---- Display results ----
        print("-" * 50)
        if data.get("status_code") == 403:
            print(f"🚨 Status    : {data['status_desc']}")
            print(f"🏷️  Fraud Type : {data['fraud_type']}")
            print(f"📝 Details   : {data['fraud_desc']}")
        else:
            print("✅ Message appears SAFE.")
        print("-" * 50)

        return data

    except requests.exceptions.ConnectionError:
        print("❌ Error: Could not connect to the API. Check your internet or the API URL.")
    except requests.exceptions.HTTPError as e:
        print(f"❌ HTTP Error: {e}")
    except Exception as e:
        print(f"❌ Unexpected error: {e}")

    return {}
```

## ✏️ Step 4 — Enter Your Message & Run

```python
# ✏️ Change this message and re-run the cell
message = "If you want to become rich, transfer N1,000 now"

result = predict(message)
```

## 🔁 Step 5 — Batch Check Multiple Messages

```python
# Test multiple messages at once
messages = [
    "If you want to become rich, transfer N1,000 now",
    "Your account has been compromised, click here to verify",
    "Hey, are we still meeting tomorrow at 10am?",
    "Congratulations! You have won a free iPhone. Claim now!"
]

for i, msg in enumerate(messages, 1):
    print(f"\n📩 Message {i}: {msg}")
    predict(msg)
```

## 🧾 Step 6 — Access Raw Response Fields

```python
# If you want to use the result programmatically
message = "Send me your bank details to receive your prize"
result = predict(message)

# Access individual fields
if result:
    print("\nRaw fields:")
    print("status_code :", result.get("status_code"))
    print("status_desc :", result.get("status_desc"))
    print("fraud_type  :", result.get("fraud_type"))
    print("fraud_desc  :", result.get("fraud_desc"))
```
